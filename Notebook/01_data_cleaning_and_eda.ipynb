### Business Rule: Missing Return Information

**Dataset:** `ecommerce_sales_customer_analytics_150k.csv`

**Columns:** `order_status`, `return_status`, `return_reason`

**Observation:** Approximately 93.15% of `return_status` values are missing. Missing values appear to be associated with the order lifecycle.

**Decision:** Do not replace missing return statuses until the business meaning has been validated. If the data definition confirms that missing means the order was not returned, classify those records as `Not Returned`. Otherwise, preserve null values for further investigation.

**Reason:** Prevents incorrect assumptions and maintains data integrity.

**Validation status:** Pending confirmation of the business rule.

## Data Type Validation and Correction

After inspecting the dataset using `ecommerce_df.info()`, the following columns require data type validation or correction.

| Column | Current Data Type | Recommended Data Type | Reason |
|---|---|---|---|
| `order_date` | `str` | `datetime64[ns]` | Enables date filtering, monthly/yearly analysis, and time-series reporting. |
| `order_time` | `str` | `str` or time-compatible type | A string is acceptable if used only for display; convert if performing time-based analysis. |
| `customer_postal_code` | `int64` | `str` | Postal codes are identifiers, not quantities. String storage preserves leading zeros and supports alphanumeric codes. |

### Data Type Decisions

- **`order_date`:** Convert to a datetime type for accurate date operations.
- **`order_time`:** Keep as a string if its format is consistent and no time calculations are required. Otherwise, parse it into a suitable time representation.
- **`customer_postal_code`:** Convert to a string because mathematical operations on postal codes are not meaningful.

### Validation Principle

Other columns should retain their current types unless profiling identifies a specific problem. For example, `customer_id` and `order_id` are already strings, while sales, cost, profit, and quantity columns have numeric types suitable for calculations.

```python
ecommerce_df["order_date"] = pd.to_datetime(
    ecommerce_df["order_date"],
    errors="coerce"
)
ecommerce_df['order_time'] = pd.to_datetime(ecommerce_df['order_time'], errors='coerce')
ecommerce_df['customer_postal_code'] = ecommerce_df['customer_postal_code'].astype(str)


ecommerce_df.info()
```

```python
fig, axes = plt.subplots(3, 3, figsize=(20, 20))
axes = axes.flatten()
columns = ["shipping_method", "payment_method", "payment_status"]

for ax, col in zip(axes, columns):
    missing_pct = (
        ecommerce_df
        .groupby(col, dropna=False)["return_status"]
        .apply(lambda x: x.isna().mean() * 100)
        .reset_index(name="missing_percentage")
    )
    sns.barplot(data=missing_pct, x=col, y="missing_percentage", ax=ax)
    ax.set_title(f"Missing return_status by {col}")

for ax in axes[len(columns):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
ecommerce_df['return_status'] = ecommerce_df['return_status'].fillna('Not Returned')
ecommerce_df['return_reason'] = ecommerce_df['return_reason'].fillna('Not Applicable')
```

Handling Return-Related Missing Values
The missing values in return_status and return_reason were found to be structurally related. All orders marked as returned had a valid return reason, while orders without a return had missing values in both fields.

Therefore:

Missing return_status values were replaced with "Not Returned"
Missing return_reason values were replaced with "Not Applicable"
These values were not treated as data errors because the information does not apply to non-returned orders.

Delivery Days Missing Values
All missing values in delivery_days correspond to orders with a delivery_status of "Cancelled". Since these orders were not successfully delivered, an actual delivery duration does not exist.

Therefore, the missing values in delivery_days were kept as NaN rather than being replaced with an artificial numeric value such as 0.

```python
(
    ecommerce_df['delivery_days'].isnull()
    !=
    ecommerce_df['estimated_delivery_days'].isnull()
).sum()
```

Estimated Delivery Days Missing Values
The missing values in estimated_delivery_days occur in exactly the same rows as the missing values in delivery_days. These rows correspond to orders with a delivery status of "Cancelled".

Therefore, the missing values were kept as NaN because no delivery duration was recorded for cancelled orders.
