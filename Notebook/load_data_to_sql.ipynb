```python

import os
import pandas as pd
import sqlalchemy
from sqlalchemy import create_engine
from pathlib import Path


db_user = 'root'
db_password = 'your_password'
db_host = 'localhost'
db_port = '3306'
db_name = 'your_database_name'

connection_string = f'mysql+pymysql://{db_user}:{db_password}@{db_host}:{db_port}/{db_name}'
engine = create_engine(connection_string)

print("Database connection established successfully!")


data_path =  Path.cwd().parent / 'data' / 'data_modeling'  


dim_customer = pd.read_csv(os.path.join(data_path, 'dim_Customer.csv'))
dim_product  = pd.read_csv(os.path.join(data_path, 'dim_Product.csv'))
dim_date     = pd.read_csv(os.path.join(data_path, 'dim_Date.csv'))
fact_orders  = pd.read_csv(os.path.join(data_path, 'fact_Orders.csv'))
fact_sales   = pd.read_csv(os.path.join(data_path, 'fact_Sales.csv'))

print("All CSV files loaded into DataFrames successfully!")


print("Uploading Dimension tables...")
dim_customer.to_sql('Customers', con=engine, if_exists='append', index=False)
dim_product.to_sql('Products', con=engine, if_exists='append', index=False)
dim_date.to_sql('Dates', con=engine, if_exists='append', index=False)
print("Dimension tables uploaded.")

print("Uploading Fact tables...")
fact_orders.to_sql('Fact_Orders', con=engine, if_exists='append', index=False)
fact_sales.to_sql('Fact_sales_transactions', con=engine, if_exists='append', index=False)
print("Fact tables uploaded.")

print("All tables successfully populated in MySQL!")
```
